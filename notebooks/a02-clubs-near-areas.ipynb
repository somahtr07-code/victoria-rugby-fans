# A02: Rugby union clubs near the top areas (Victoria)
This notebook maps Victoria's rugby union clubs, and later checks which of the top 10 areas from A01 have a club nearby.

## 1. Load the club table
26 clubs from Rugby Victoria's 2026 competition pages, with home-ground coordinates from Google Maps.

```python
import pandas as pd
import os
for root, dirs, files in os.walk("/kaggle/input"):
    print(root, files)
```

```python
club_path = "/kaggle/input/datasets/somahattori/victoria-rugby-clubs-2026/"
clubs = pd.read_csv(club_path + "rugby_clubs_vic.csv")
clubs = clubs[clubs["include_in_a02"] == "yes"].copy()
print(clubs.shape)
clubs[["club", "home_ground", "lat", "lon"]].head()
```

## 2. Map the clubs
Each dot is one club's home ground. Hover over a dot to see the club name.

```python
import plotly.express as px
fig = px.scatter_mapbox(clubs, lat="lat", lon="lon", hover_name="club",
                        hover_data=["home_ground"], zoom=8.5,
                        center={"lat": -37.95, "lon": 145.0}, height=600)
fig.update_layout(mapbox_style="open-street-map", margin=dict(l=0, r=0, t=0, b=0))
fig.show()
```

**What the map shows (by eye, before calculating distances)**
- Most clubs are in inner and middle Melbourne, especially in the inner south and the south-east.
- There are fewer clubs in the north and the west of Melbourne.
- There is no club on the Mornington Peninsula. The southernmost clubs in Melbourne are Southern Districts and Casey. Four of the top 10 areas from A01 are on the peninsula (Mornington West, Mornington East, Mount Eliza and Mount Martha).
- Outside Melbourne, there are clubs in Geelong and Melton on this map. Ballarat, Bendigo and Shepparton are outside this view (zoom out to see them).
